```python
import pandas as pd
df = pd.read_excel('C:/Place_for_training/project_3/Crime_Data.xlsx')
df
```

```python
df_copy = df.copy()
df_copy
```

```python
df_copy.info()
```

```python
df_copy.columns = df_copy.columns.str.replace(" ","_")
df_copy.columns
```

```python
df_copy['Ward'] = df_copy.groupby('Beat')['Ward'].transform(
    lambda x:x.fillna(x.mode()[0] if not x.mode().empty else pd.NA))
df_copy['Community_Area'] = df_copy['Community_Area'].fillna(df_copy.groupby('Beat')['Community_Area'].transform(
    lambda x: x.mode()[0] if not x.mode().empty else pd.NA)
)
df_copy.dropna(subset=['Ward', 'Community_Area', 'Latitude', 'Location_Description'], inplace=True)
df_copy['Ward'] = df_copy['Ward'].astype(int)
df_copy['Community_Area'] = df_copy['Community_Area'].astype(int)
df_copy = df_copy.drop(columns='Location')
df_copy.info()
```

```python
df_copy = df_copy[df_copy['Latitude'] > 40]
df_copy.info()
```

```python
df_copy.to_csv('Cleaned_Chicago_Crime.csv', index=False)
```
